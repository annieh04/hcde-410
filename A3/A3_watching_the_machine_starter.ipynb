## 4. Synthesis & reflection

*Bring the probe and the observation together into one account. What does this mean for trustworthy, reproducible research — and what will you change about how you use and verify AI in your own work?*

The user chose to check the spreadsheet manually after getting the answer from Gemini. This suggests that for tasks where manual verification is reasonable for the user, they would choose to double check rather than fully trust the LLM's answer. I would say I have the same habit as well. In scenarios where manual verification is unreasonable, I would generally half to trust the LLM's answer. If the answer is extremely important, I might ask it to double check and/or would probably put the same prompt into a different LLM to verify if it gives the same answer, it which case I would be able to trust more that the answer given is correct. This cross-checking is a technique I've heard people using and is something I might consider doing more often, when using LLM for more important tasks. 

My observation made an interesting case for a lack of reproducibility, even when starting from the same prompt. Because the user chose to clarify the ambiguous criteria of "summer" and I did not, we had different correct answers. This also demonstrates how the assumptions made by AI are important to watch out for. In the initial responses given to me by Gemini, it gave 2 options for different categorizations of what I might consider the summer months to be. However, in later responses, it opted for concision and chose to make an assumption about which categorization I would want. With a different scenario, this could lead to a fatal error. Thus, it is important to recognize areas in your prompt where AI might make assumptions about what you mean. For my own personal usage of AI, I should define key terms rather than assuming AI makes the choice I want. Another way around this is to ask AI to state very clearly what assumptions are made, so that they don't get hidden.
